## Exercise 3.1 

```python
mu = 3000
samples = [3003, 3005, 2997, 3006, 2999, 2998,3007, 3005, 3001]
```

a)

Compute the following three statistics, mean. The sample standard deviation, and the standard error of the mean.

```python
arr = np.array(samples)
n = len(samples)
mean = arr.mean()
std = arr.std(ddof=1)
stderr = std/n**0.5

print(stderr)

```

b) 

In a construction process, 5 concrete items are joined together to a single construction with a length which is then the complete length of the 5 concrete items. It is very important that the length of this new construction is within 15 m plus/minus 1 cm. How often will it happen that such a construction will be more than 1 cm away from the 15 m target (assume that the population mean concrete item length is mu = 3000 mm and that the population standard deviation is sigma = 3)

Sum of independent random variables

$ T = X_{1} + X_{2} + X_{3} + X_{4} + X_{5} $

$ E[T] = 5 \mu = 5 \cdot 3000 = 15000 \mathrm{mm} = 15 \mathrm{m}$

$ \mathbf{Var}(T) =5 \sigma^{2} = 5 \cdot 3^{2} = 45 \mathrm{mm}^{2} $ 

$ SD(T) = \sqrt{45} \approx 6.708 \mathrm{mm}$ 




c) find the 95% confidence interval for the mean $\mu$

```python
df = n - 1 

t_crit = stats.t.ppf(0.975, df)

margin = t_crit * stderr
print(f"margin: {margin}")

ci = (mean - margin, mean + margin)

print(f"Confidence interval {ci[0]}-{ci[1]}")
```

```python
var = arr.var()
conf_95 = 0.95
conf_99 = 0.99

alpha = 1 - conf_95

chi2_lower = stats.chi2.ppf(alpha/2, df)
chi2_upper = stats.chi2.ppf(1 - alpha/2, df)


var_low  = (df * std**2) / chi2_upper
var_high = (df * std**2) / chi2_lower

sd_low, sd_high = np.sqrt(var_low), np.sqrt(var_high)

print(f"Confidence interval var: [{var_low} < {var} < {var_high}]")
print(f"Confidence interval std: [{sd_low} < {std} < {sd_high}]")
```
