```python
n = 9
firm_a_mean = 1.93
firm_b_mean = 1.49

s_a = 0.45 
s_b = 0.58
```

$ H_{0}: \mu_{A} = \mu_{B} $
 
a) What is the p-value, and conclusion for this test at $\alpha = 5\%$ 

```python
s_p = np.sqrt(((n - 1) * s_a**2 + (n-1)*s_b**2)/(n+n-2))

T = (firm_a_mean - firm_b_mean)/(s_p * np.sqrt(1/n + 1/n ))
df = n+n-2
test = stats.t.ppf(1-0.05/2, n+n-2)

print(f"Reject null hypothesis: {T > test}")
```
