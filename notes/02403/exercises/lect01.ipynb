## Exercise 1.1

```python
x = [2474, 2547, 2830, 3219, 3429, 3448, 3677, 3872, 4001, 4116] 
y = [2844, 2863, 2963, 3239, 3379, 3449, 3582, 3926, 4151, 4356]

def mean(x):
    return sum(x) / len(x) 

x_mean = mean(x)

def variance(data, mean):
    s = 0
    for i in data:
        s += (i - mean)**2

    return s / (len(data) - 1)

var = variance(x, x_mean)
x_std = var**0.5

print(x_mean)
print(var)
print(x_std)
```

b)

```python
y_mean = mean(y)
y_var = variance(y, y_mean)
y_std = y_var**0.5

print(y_mean, y_var, y_std)
```

## Exercise 1.1

```python
height = np.array([168, 161, 167, 179, 184, 166, 198, 187, 191, 179])
weight = np.array([65.5, 58.3, 68.1, 85.7, 80.5, 63.4, 102.6, 91.4, 86.7, 78.9])

sigma = np.array([[np.std(height, ddof=1), 0], [0, np.std(weight, ddof=1)]  ])
R = np.corrcoef(height,weight)

print(sigma)

S = np.cov(height, weight, ddof=1)

print(S == sigma @ R @ sigma)
print(S)
```

## Exercise 1.2

```python
eigenvalues, eigenvectors = np.linalg.eig(S)

D = np.diag(eigenvalues)
P = eigenvectors
P_inv = np.linalg.inv(eigenvectors)

S_half = P @ D**0.5 @ P.T

S_half @ S_half.T
```
